### 1. Buktikan dengan tabel kebenaran bahwa ¬(p ∧ q) ekuivalen dengan ¬p ∨ ¬q (hukum De Morgan)

| p | q | p ∧ q | ¬(p ∧ q) | ¬p | ¬q | ¬p ∨ ¬q |
|---|---|---|---|---|---|---|
| B | B | B | S | S | S | S |
| B | S | S | B | S | B | B |
| S | B | S | B | B | S | B |
| S | S | S | B | B | B | B |

Kolom **¬(p ∧ q)** dan **¬p ∨ ¬q** sama-sama bernilai **S, B, B, B**, sehingga terbukti:

$$ \neg(p \land q) \equiv \neg p \lor \neg q $$

### 2. Terjemahkan ke logika predikat

Misal:
- $M(x)$: $x$ adalah mahasiswa.
- $P(x)$: $x$ adalah praktikum.
- $L(x)$: $x$ dikumpulkan lewat LMS.
- $K(x,y)$: $x$ mengumpulkan $y$.
- $T(x)$: $x$ mengumpulkan tepat waktu.
- $N(x)$: $x$ mendapat nilai penuh.

**(a) “Semua praktikum dikumpulkan lewat LMS.”**

$$ \forall x (P(x) \rightarrow L(x)) $$

**(b) “Ada mahasiswa yang belum mengumpulkan P05.”**

$$ \exists x (M(x) \land \neg K(x, P05)) $$

**(c) “Setiap mahasiswa yang mengumpulkan tepat waktu mendapat nilai penuh.”**

$$ \forall x ((M(x) \land T(x)) \rightarrow N(x)) $$

### 3. Lima aturan produksi untuk domain jaringan WiFi kos

1. **R1:** JIKA tidak ada SSID terdeteksi DAN lampu indikator router mati MAKA router tidak mendapat daya.  
2. **R2:** JIKA router tidak mendapat daya MAKA catu daya/adaptor bermasalah.  
3. **R3:** JIKA catu daya/adaptor bermasalah MAKA router mati.  
4. **R4:** JIKA router mati MAKA semua perangkat tidak bisa terhubung ke WiFi.  
5. **R5:** JIKA perangkat terhubung ke WiFi tetapi tidak bisa internet DAN router menyala MAKA uplink/ISP bermasalah.

Rantai aturan:  
**R1 → R2 → R3 → R4**  
Kesimpulan R1 menjadi kondisi R2, kesimpulan R2 menjadi kondisi R3, dan seterusnya.

### 4. Jaringan semantik minimal 6 simpul untuk dunia perkuliahan

    [Orang] --memiliki--> [Nama]
       ^ is-a               ^ is-a
       |                    |
    [Mahasiswa]          [Dosen]
       | terdaftar di       | mengajar
       v                    v
    [Prodi] <--diselenggarakan oleh-- [MataKuliah]
       | berada di
       v
    [Gedung]

Panah berlabel:
- Mahasiswa **is-a** Orang.
- Dosen **is-a** Orang.
- Mahasiswa **terdaftar di** Prodi.
- Dosen **mengajar** MataKuliah.
- MataKuliah **diselenggarakan oleh** Prodi.
- Prodi **berada di** Gedung.
- Orang **memiliki** Nama.

Contoh pewarisan: karena **Mahasiswa is-a Orang**, maka Mahasiswa otomatis mewarisi sifat **memiliki nama** dari Orang. Begitu pula Dosen.

### 5. Kasus penguin: penguin is-a burung tetapi tidak bisa terbang

Representasi frame dapat menangani pengecualian dengan **default inheritance** dan **override**.

**Frame Burung**
- is-a: Hewan
- bersayap: ya
- bertelur: ya
- bisa_terbang: **default ya**
- pengecualian: Penguin

**Frame Penguin**
- is-a: Burung
- bisa_terbang: **tidak**
- habitat: dingin
- warna: hitam-putih

Mekanisme:
- Penguin mewarisi sifat **bersayap** dan **bertelur** dari Burung.
- Namun, slot **bisa_terbang** di frame Penguin menimpa default dari frame Burung.
- Jadi, Penguin tetap “is-a burung”, tetapi tidak bisa terbang.

#Pengembangan

```python
import numpy as np                 # np = nama panggilan standar
import pandas as pd                # pd = nama panggilan standar

# ============================================================
# 1. MEMBACA DATA DARI FILE CSV
# ============================================================
# Jika di Google Colab, upload dulu file "nilai_mahasiswa (1).csv"
# lalu sesuaikan nama file di bawah ini.
df = pd.read_csv("nilai_mahasiswa.csv")

print("=== 5 Data Pertama ===")
print(df.head())

print("\n=== Info Dataset ===")
print(df.info())

print("\n=== Statistik Deskriptif ===")
print(df.describe())

# ============================================================
# 2. HITUNG NILAI AKHIR
#    Rumus: 40% UTS + 50% UAS + 10% Kehadiran
# ============================================================
df["nilai_akhir"] = 0.4 * df["uts"] + 0.5 * df["uas"] + 0.1 * df["kehadiran"]

# ============================================================
# 3. STATISTIK NILAI AKHIR
# ============================================================
print("\n=== Statistik Nilai Akhir ===")
print("Rata-rata   :", df["nilai_akhir"].mean())
print("Tertinggi   :", df["nilai_akhir"].max())
print("Terendah    :", df["nilai_akhir"].min())
print("Std Deviasi :", df["nilai_akhir"].std())

# ============================================================
# 4. MENAMBAHKAN KOLOM PREDIKAT (GRADE)
# ============================================================
def predikat(nilai):
    if nilai >= 85:
        return "A"
    elif nilai >= 75:
        return "B"
    elif nilai >= 65:
        return "C"
    elif nilai >= 55:
        return "D"
    else:
        return "E"

df["predikat"] = df["nilai_akhir"].apply(predikat)

print("\n=== Tabel Lengkap dengan Nilai Akhir & Predikat ===")
print(df[["nim", "nama", "kelas", "uts", "uas", "kehadiran", "nilai_akhir", "predikat"]])

# ============================================================
# 5. MAHASISWA DENGAN NILAI TERTINGGI & TERENDAH
# ============================================================
print("\n=== Mahasiswa Nilai Tertinggi ===")
print(df.loc[df["nilai_akhir"].idxmax()])

print("\n=== Mahasiswa Nilai Terendah ===")
print(df.loc[df["nilai_akhir"].idxmin()])

# ============================================================
# 6. ANALISIS PER KELAS
# ============================================================
print("\n=== Rata-rata Nilai Akhir per Kelas ===")
print(df.groupby("kelas")["nilai_akhir"].mean())

print("\n=== Jumlah Mahasiswa per Kelas ===")
print(df.groupby("kelas")["nim"].count())

# ============================================================
# 7. DISTRIBUSI PREDIKAT
# ============================================================
print("\n=== Distribusi Predikat per Kelas ===")
tabel = pd.crosstab(df["kelas"], df["predikat"])
print(tabel)

# ============================================================
# 8. URUTKAN DARI NILAI TERTINGGI
# ============================================================
print("\n=== Ranking Mahasiswa (Nilai Tertinggi ke Terendah) ===")
df_urut = df.sort_values("nilai_akhir", ascending=False)
df_urut["ranking"] = range(1, len(df_urut) + 1)
print(df_urut[["ranking", "nim", "nama", "kelas", "nilai_akhir", "predikat"]])

# ============================================================
# 9. FILTER: MAHASISWA YANG PERLU REMEDIAL (nilai < 60)
# ============================================================
print("\n=== Mahasiswa Perlu Remedial (Nilai Akhir < 60) ===")
remedial = df[df["nilai_akhir"] < 60]
print(remedial[["nim", "nama", "kelas", "nilai_akhir"]])

# ============================================================
# 10. SIMPAN HASIL KE CSV BARU
# ============================================================
df.to_csv("hasil_nilai_mahasiswa.csv", index=False)
print("\n✅ Hasil disimpan ke 'hasil_nilai_mahasiswa.csv'")
```

#Pengembangan

```python
import pandas as pd

# ============================================================
# 1. MUAT DATA
# ============================================================
df = pd.read_csv("nilai_mahasiswa.csv")   # muat CSV jadi DataFrame
print(df.head())                     # intip 5 baris pertama
print(df.info())                     # tipe kolom + cek data kosong
print(df.describe())                 # statistik semua kolom angka

# ============================================================
# 2. KOLOM BARU: NILAI AKHIR
# ============================================================
# kolom baru: nilai akhir = 40% UTS + 60% UAS (vektorisasi ala NumPy)
df["nilai_akhir"] = 0.4 * df["uts"] + 0.6 * df["uas"]

# tambahan: selisih UTS -> UAS & rata-rata mentah
df["selisih"]   = df["uas"] - df["uts"]
df["rata_rata"] = (df["uts"] + df["uas"]) / 2

# ============================================================
# 3. FILTER BARIS
# ============================================================
pintar = df[df["nilai_akhir"] > 80]     # filter baris
print("\n=== Mahasiswa Pintar (nilai_akhir > 80) ===")
print(pintar[["nama", "kelas", "nilai_akhir"]])

remedial = df[df["nilai_akhir"] < 60]
print("\n=== Mahasiswa Remedial (nilai_akhir < 60) ===")
print(remedial[["nim", "nama", "kelas", "nilai_akhir"]])

rajin = df[df["kehadiran"] >= 90]
print("\n=== Mahasiswa Rajin (kehadiran >= 90) ===")
print(rajin[["nama", "kelas", "kehadiran"]])

# ============================================================
# 4. GROUPBY: RATA-RATA PER KELAS
# ============================================================
print("\n=== Rata-rata Nilai Akhir per Kelas ===")
print(df.groupby("kelas")["nilai_akhir"].mean())

print("\n=== Statistik Lengkap per Kelas ===")
print(df.groupby("kelas").agg(
    jumlah     = ("nim", "count"),
    rata_uts   = ("uts", "mean"),
    rata_uas   = ("uas", "mean"),
    rata_akhir = ("nilai_akhir", "mean"),
    max_akhir  = ("nilai_akhir", "max"),
    min_akhir  = ("nilai_akhir", "min"),
))

# ============================================================
# 5. RANKING
# ============================================================
df_rank = df.sort_values("nilai_akhir", ascending=False).reset_index(drop=True)
df_rank.index += 1                      # mulai dari ranking 1
print("\n=== Ranking 10 Besar Mahasiswa ===")
print(df_rank.head(10)[["nama", "kelas", "nilai_akhir"]])

print("\n=== Top 3 per Kelas ===")
top3 = df.sort_values("nilai_akhir", ascending=False).groupby("kelas").head(3)
print(top3[["kelas", "nama", "nilai_akhir"]])

# ============================================================
# 6. KOLOM GRADE / PREDIKAT
# ============================================================
def grade(n):
    if n >= 85:   return "A"
    elif n >= 75: return "B"
    elif n >= 65: return "C"
    elif n >= 55: return "D"
    else:         return "E"

df["grade"] = df["nilai_akhir"].apply(grade)

print("\n=== Distribusi Grade ===")
print(df["grade"].value_counts().sort_index())

print("\n=== Grade per Kelas ===")
print(pd.crosstab(df["kelas"], df["grade"]))

# ============================================================
# 7. PIVOT TABLE
# ============================================================
print("\n=== Pivot: Rata-rata Nilai Akhir (Kelas x Grade) ===")
print(df.pivot_table(
    values  = "nilai_akhir",
    index   = "kelas",
    columns = "grade",
    aggfunc = "mean"
))

# ============================================================
# 8. KORELASI
# ============================================================
print("\n=== Korelasi Antar Nilai ===")
print(df[["uts", "uas", "kehadiran", "nilai_akhir"]].corr())

# ============================================================
# 9. SIMPAN HASIL
# ============================================================
df.to_csv("hasil_nilai_mahasiswa.csv", index=False)
print("\n✅ File 'hasil_nilai_mahasiswa.csv' berhasil disimpan.")
```

```python
# Korelasi Hours_Studied vs Exam_Score
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("StudentPerformanceFactors.csv")
corr = df['Hours_Studied'].corr(df['Exam_Score'])
print(f"Korelasi Hours_Studied vs Exam_Score: {corr:.3f}")

# Scatter plot + garis regresi
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='Hours_Studied', y='Exam_Score', scatter_kws={'alpha': 0.5})
plt.title('Hubungan Hours_Studied dan Exam_Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.show()
```

Temuan 2
Jika korelasi positif, maka semakin banyak jam belajar cenderung semakin tinggi nilai ujian. Besarnya korelasi menunjukkan kekuatan hubungan tersebut. Garis regresi membantu melihat arah hubungan secara visual.

```python
# Rata-rata Exam_Score per Parental_Involvement
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("StudentPerformanceFactors.csv")
avg_parent = df.groupby('Parental_Involvement')['Exam_Score'].mean().sort_values()
print(avg_parent)

# Bar plot
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x='Parental_Involvement',
    y='Exam_Score',
    order=['Low', 'Medium', 'High'],
    errorbar=None
)
plt.title('Rata-rata Exam_Score berdasarkan Parental Involvement')
plt.xlabel('Parental Involvement')
plt.ylabel('Rata-rata Exam Score')
plt.show()
```

Temuan 3
Grafik ini menunjukkan apakah tingkat keterlibatan orang tua (Low, Medium, High) berasosiasi dengan perbedaan rata-rata nilai ujian. Jika batang High lebih tinggi, ada indikasi bahwa keterlibatan orang tua yang lebih tinggi cenderung diikuti nilai ujian yang lebih baik.
